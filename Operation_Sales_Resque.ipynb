````python
import csv
import random
from datetime import datetime, timedelta

# 1. Define our data structure (Headers)
headers = ["InvoiceID", "Date", "Product", "Quantity", "UnitPrice", "CustomerID", "Country"]

# 2. Prepare sample data pools
products = ["Notebook", "Pen", "Mug", "USB Drive", "Headphones", "Keyboard", "Mouse"]
countries = ["UK", "Germany", "France", "USA", "Spain"]

# 3. Function to generate random date
def random_date(start, end):
    return start + timedelta(days=random.randint(0, (end - start).days))

# 4. Generate 1000 rows
data_rows = []
start_date = datetime(2023, 1, 1)
end_date = datetime(2023, 12, 31)

for i in range(1000):
    invoice = f"INV-{1000 + i}"
    date = random_date(start_date, end_date).strftime("%Y-%m-%d")
    product = random.choice(products)
    quantity = random.randint(1, 10)
    price = round(random.uniform(5.0, 50.0), 2)
    customer = f"CUST-{random.randint(100, 999)}"
    country = random.choice(countries)

    data_rows.append([invoice, date, product, quantity, price, customer, country])

# 5. Write to CSV file
filename = "sales_data.csv"
with open(filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(data_rows)

print(f"✅ Success! Generated '{filename}' with {len(data_rows)} records.")
````

#🕵️‍♂️ Mission Update:
Detective, imagine you have to calculate the Risk Score for 1,000 transactions. Would you copy-paste the same if/else logic 1,000 times? Absolutely not. That's error-prone and messy.
In a police precinct, we have Standard Operating Procedures (SOPs). Everyone follows the same steps to process evidence.
In Python, an SOP is called a Function. It allows us to write the logic once, name it, and reuse it whenever we need it. This is the DRY Principle (Don't Repeat Yourself).

# 📋 Define the function

````python
def calculate_risk_score(amount):
    """Calculates risk based on transaction amount."""
    if amount > 10000:
        return "HIGH"
    elif amount > 5000:
        return "MEDIUM"
    else:
        return "LOW"
````

# 🖨️ Use the function

````python
status = calculate_risk_score(15000)
print(f"Transaction Risk: {status}")
````

# 📋 Function with two inputs

````python
def check_suspect(name, risk_level):
    if risk_level == "HIGH":
        return f"⚠️ Investigate {name} immediately!"
    else:
        return f"✅ {name} is clear."
````

# 🖨️ Call with arguments

````python
print(check_suspect("Alex Mercer", "HIGH"))
print(check_suspect("John Doe", "LOW"))
````

#🕵️‍♂️ Mission Update:
Detective, you have mastered the tools of the trade. Now it's time for your first field test.
We have received a batch of transaction records from the finance department. They suspect some are fraudulent.

# 📋 Raw transaction data (List of Dictionaries)

````python
transactions = [
    {"id": 101, "amount": 500, "merchant": "Coffee Shop"},
    {"id": 102, "amount": 15000, "merchant": "Electronics Inc"},
    {"id": 103, "amount": 200, "merchant": "Grocery"},
    {"id": 104, "amount": 25000, "merchant": "Luxury Cars"},
    {"id": 105, "amount": 8000, "merchant": "Furniture"}
]
````

# 📋 Function to check single transaction

````python
transaction
def is_flagged(transaction):
    if transaction["amount"] > 10000:
        return True
    else:
        return False
````

# 🔄 Loop and Filter

````python
flagged_transactions = []

for tx in transactions:
    if is_flagged(tx):
        flagged_transactions.append(tx)
````

# 🖨️ Report Results

````python
print(f"🚨 Found {len(flagged_transactions)} suspicious transactions:")
for item in flagged_transactions:
    print(f"ID: {item['id']} | Amount: ${item['amount']}")
````
